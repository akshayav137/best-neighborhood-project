```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("median age.csv")

neighborhood_col = "NEIGHBORHOOD"
median_age_col = "TOTAL MD AGE AT DEATH"

# removes rows that have missing media/n values
df = df.dropna(subset=[median_age_col])

# 🔹 Sort by ascending order
df = df.sort_values(by=median_age_col, ascending=True)

# Plot
plt.figure(figsize=(12, 6))
plt.bar(df[neighborhood_col], df[median_age_col], color = "pink")

plt.title("Median Age at death by Neighborhood")
plt.xlabel("Neighborhood")
plt.ylabel("Total Median Age")
plt.xticks(rotation=90)
plt.show()
```
